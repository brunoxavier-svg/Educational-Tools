```python
import numpy as np
import plotly.graph_objects as go

L = float(input('Informe o comprimento L da corda: '))
v = float(input('Informe a velocidade v da onda: '))

steps = []
for lamb in np.arange(0.5, 10, 0.1):

  f = v/lamb
  n = 2*L/lamb

  x = np.linspace(0, L, 1000)

  # Onda original
  y1 = np.sin(n*np.pi*x/L)

  # Onda refletida com 0.6 de amplitude
  y2 = 0.6*np.sin(n*np.pi*x/L)

  steps.append(dict(
      method="restyle",
      args=[{"x": [x, x],
             "y": [y1, y2]},
            {"name": f'λ = {lamb:.2f} m'}],
      label=f'λ = {lamb:.2f} m'))

fig = go.Figure(data=[go.Scatter(x=[0], y=[0], visible=True)])

fig.update_traces(line_color='blue', selector=dict(name="Onda original"))
fig.add_trace(go.Scatter(x=[0], y=[0], visible=True,
                         line=dict(color='red', width=3),
                         name='Onda refletida'))

# Layout
fig.update_layout(
    title = "Onda estacionária",
    xaxis_title = "Posição (m)",
    yaxis_title = "Amplitude",
    sliders = [dict(
         active = 0,
         steps = steps
    )]
)

fig.show()
```
